```python
def convert(obj):
    L = []
    for i in ast.literal_eval(obj):
        L.append(i['name'])
    return L
#Hàm convert(obj): Chuyển một chuỗi biểu diễn danh sách các đối tượng (dict) có khóa 'name' thành một danh sách chỉ chứa các giá trị 'name'.
```

```python
movies['genres'] = movies['genres'].apply(convert)
```

```python
movies['genres']
```

```python
movies['keywords'] = movies['keywords'].apply(convert)
```

```python
movies['keywords']
```

```python
movies['cast'] = movies['cast'].apply(lambda x: [i['name'] for i in ast.literal_eval(x)[:3]])  
# Chuyển chuỗi JSON đó thành danh sách thật của các dict Python
# Only top 3 actors
# Tạo danh sách chỉ chứa tên diễn viên
```

```python
movies['crew'] = movies['crew'].apply(lambda x: [i['name'] for i in ast.literal_eval(x) if i['job'] == 'Director'])
```

```python
movies['tags'] = movies['genres'] + movies['keywords'] + movies['cast'] + movies['crew']

# tag của movie là gồm các cột genres, keywwords, cast, crew
```

```python
movies['tags'] = movies['tags'].apply(lambda x: " ".join(x)) #join lại bằng dấu cách
```

```python
class MyTfidfVectorizer:
    def __init__(self):
        self.vocab_ = {} # một dict tạo ánh xạ: từ -> index
        self.idf_ = None # vector idf dc tính sau

    def _tokenize(self, doc: str):
        # tags của bạn dạng "Action Adventure Tom_Hanks ..." nên split theo space là đủ
        return doc.lower().split()
    # chuyển thành chữ thường và tách theo khoảng trắng, mỗi từ dc tách ra gọi là 1 token, output sẽ là 1 list các từ đơn được tách ra

    def fit_transform(self, docs):
        """
        docs: iterable (list/Series) các chuỗi.
        Trả về: ma trận TF-IDF dạng numpy.ndarray shape (n_docs, n_terms)
        """
        # 1. Tokenize từng doc
        tokenized_docs = [self._tokenize(d) for d in docs]

# tokenized_docs sẽ là ma trận 2 chiều như này
#[
#    ["action", "adventure", "tom_hanks"],      # mỗi dòng này là tag của từng phim, mà tag của từng phim lại dc split thành list các token
#    ["comedy", "romance", "jennifer_lawrence"], 
#    ["sci-fi", "action", "tom_cruise"]        
#]
        # 2. Xây vocab
        vocab = {}
        for tokens in tokenized_docs: # duyệt qua tag của từng phim 1, từng row
            for t in tokens: #trong tag của 1 phim thì lại là list các token của tag đó nên duyệt qua từng từ
                if t not in vocab: # nếu từ đấy chưa dc đánh index
                    vocab[t] = len(vocab) # thì key = từ đó, value bằng len của vocab lúc đó
        self.vocab_ = vocab

        n_docs = len(tokenized_docs) # số lượng documents (phim) trong dataset
        n_terms = len(vocab) # số lượng từ khác nhau

        # 3. Tính TF (term frequency) & DF (document frequency)
        X = np.zeros((n_docs, n_terms), dtype=np.float32) #khởi tạo ma trận tf ban đầu (x), hàng là các phim, cột là các từ khác nhau
        df_counts = np.zeros(n_terms, dtype=np.int32) #tạo mảng df bản đầu (1d) 
        for i, tokens in enumerate(tokenized_docs): # i là stt phim thứ i, tokens là list các từ đã tách của phim i
            counts = Counter(tokens) # đếm tần suất từ trong phim, trả về 1 dict key là từ, value là tần suất 
            #{'action': 1, 'adventure': 1, 'tom_hanks': 1}
            for t, c in counts.items(): #t là key, c là value
                j = vocab[t] #lấy index của từ trong vocab
                X[i, j] = c  #gán tf 
                df_counts[j] += 1 #tăng df
#tf là từ xuất hiện bao nhiêu lần trong 1 phim (một từ xhien nhiều trong 1 phim thì qtrong với phim đó)
#df là từ xuất hiện trong bao nhiêu phim (nhưng nếu từ đó xhien trong nhiều phim thì sẽ trở nên ít quan trọng)
        # 4. Tính IDF: giống công thức phổ biến của sklearn
        # idf_j = log((1 + n_docs) / (1 + df_j)) + 1
        idf = np.log((1.0 + n_docs) / (1.0 + df_counts)) + 1.0
        self.idf_ = idf

        # 5. TF-IDF = TF * IDF
        X *= idf

        # (Có thể chuẩn hóa theo độ dài doc nếu muốn giống sklearn hơn,
        # nhưng ở đây để đơn giản, chuẩn hóa sẽ làm ở bước cosine)
        return X
```

```python
def cosine_similarity_custom(X: np.ndarray) -> np.ndarray:
    """
    X: ma trận (n_samples, n_features)
    Trả về: ma trận similarity (n_samples, n_samples)
    """
    # Chuẩn hóa từng vector hàng về norm = 1
    norms = np.linalg.norm(X, axis=1, keepdims=True) # tính norm theo từng hàng, giữ nguyên số chiều
    # Tránh chia cho 0
    norms = np.where(norms == 0, 1e-9, norms)
    #nếu norms = 0 thì thay bằng 0.00000..1, còn không thì giữ nguyên
    X_norm = X / norms # tính X_norm

    # Cosine similarity = X_norm * X_norm.T
    return X_norm @ X_norm.T #tính tích vô hướng
```

```python
tfidf = MyTfidfVectorizer() #tạo object từ class
vectors = tfidf.fit_transform(movies["tags"].values) 
    #gọi method fit_transform trên object tfidf, truyền data vào là cột tag
    # .values để đảm bảo là list/array các chuỗi
    #vectors là ma trận tf idf 
similarity = cosine_similarity_custom(vectors) 
    # tính similarity trên ma trận tf idf, output là ma trận n phim x n phim, 
    # mỗi ô [i,j] là độ tương đồng giữa phim i và j
```

```python
def get_recommendations(title, cosine_sim=similarity):
    idx = movies[movies['title'] == title].index[0] # lấy index của phim có title = title cần tìm
    sim_scores = list(enumerate(cosine_sim[idx])) # cosine_sim[index] để lấy độ tương đồng của phim idx với các phim khác. VD cosine_sim[0] = [1.0, 0.43, 0.12, 0.09, ...]
    #enumurate gắn chỉ số phim tương ứng với giá trị similarity. VD [(0, 1.0), (1, 0.43), (2, 0.12), (3, 0.09), ...]
    sim_scores = sorted(sim_scores, key=lambda x: x[1], reverse=True) #sort theo độ similarity
    sim_scores = sim_scores[1:11]  # Get top 10 similar movies
    movie_indices = [i[0] for i in sim_scores] # lấy ds chỉ số của 10 phim đó
    return movies['title'].iloc[movie_indices] # trả về title 10 phim
```

```python
print(get_recommendations('The Dark Knight Rises'))
```
